## Etapa de Consolidação das Tabelas

Em vez de manter cada tabela carregada em uma variável global separada
(`RENDIMENTO_TRABALHO`, `MORADOR`, `DOMICILIO`, ...), esta etapa consolida
todas elas em uma única estrutura, o dicionário `TABELAS_POF`
(`{nome_tabela: DataFrame}`). Cada tabela mantém sua granularidade original
(domicílio, morador, item de despesa, item de consumo alimentar etc.) — a
consolidação é estrutural, no código, e não um merge/join de linhas entre
tabelas de granularidades diferentes (o que geraria explosão combinatória de
linhas). A partir daqui, todas as etapas seguintes (tradução, perfilagem,
exportação) operam sobre esse dicionário único, em loop, em vez de repetir
código tabela por tabela.

```python
# Etapa de consolidação: reúne todas as tabelas já carregadas em uma única estrutura
TABELAS_POF = {}
for nome_tabela in ESQUEMAS_POF:
    df_tabela = globals().get(nome_tabela)
    if isinstance(df_tabela, pd.DataFrame):
        TABELAS_POF[nome_tabela] = df_tabela
    else:
        print(f"{nome_tabela}: tabela não carregada, não incluída em TABELAS_POF.")

print(f"\n{len(TABELAS_POF)}/{len(ESQUEMAS_POF)} tabelas consolidadas em TABELAS_POF:")
for nome_tabela, df_tabela in TABELAS_POF.items():
    print(f"  - {nome_tabela}: {df_tabela.shape[0]} linhas, {df_tabela.shape[1]} colunas")
```

## Etapa de Tradução Geral

Etapa única de tradução que reúne todos os tradutores construídos até aqui em um
só pipeline, aplicado de uma vez sobre todas as tabelas consolidadas em
`TABELAS_POF`:

1. **Dicionário de variáveis da POF** (`.xls`): categorias descritas
   diretamente na planilha (ex.: `"1 – Rondônia"`), incluindo o caso
   especial `"Branco – Não Aplicável"` para campos em branco.
2. **Tabelas de referência externas simples** (código → descrição, uma coluna
   `<VARIAVEL>_DESC` por variável): variáveis cujo dicionário apenas remete a
   um arquivo à parte — `ESTRATO_POF` (Estratos POF 2017-2018), `V9004`
   (Cadastro de Locais de Aquisição), `V9009` na Caderneta Coletiva (Cadastro
   de Pesos ou Volumes), `V9007` na Caderneta Coletiva (Cadastro de Unidades
   de Medida), `V53061` (Atividade CNAE Domiciliar 2.0) e `V53011` (Ocupação
   COD) no Rendimento do Trabalho, e — específicos do Consumo Alimentar —
   `V9001` (Cadastro de Produtos do Consumo Alimentar) e `V9007` /
   `COD_UNIDADE_MEDIDA_FINAL` (Cadastro de Unidades de Medidas do Consumo
   Alimentar). `V9001` no Rendimento do Trabalho e em Outros Rendimentos
   combina duas fontes complementares (Cadastro de Produtos + Tradutor de
   Rendimento), já que a faixa de código usada nessas tabelas (53xxx-55xxx)
   não está coberta pelo Cadastro de Produtos geral.
3. **Grupo de alimentos**: classificação extra (não substitui a tradução do
   produto) de `V9001` em Consumo Alimentar, em `V9001_GRUPO_DESC`
   (Classificação dos grupos de Consumo Alimentar).
4. **Dimensões de despesa**: classificação hierárquica (3 níveis) de `V9001`,
   aplicada por variável de valor deflacionado presente na tabela
   (`V8000_DEFLA`, `V8500_DEFLA`, `V8501_DEFLA`), em colunas
   `V9001_DIMENSAO_<VARIAVEL_DEFLA>` — combina as planilhas de dimensões de
   despesa (faixas de quadro 2-5 e 6-9, sem sobreposição de código).
5. **Alimentação e Aquisição Alimentar**: duas classificações extras de `V9001`
   (aplicadas em qualquer tabela que tenha essa coluna, sem depender de uma
   variável de valor deflacionado), em `V9001_ALIMENTACAO_DESC` (alimentação
   no domicílio x fora do domicílio, por grupo de alimentos) e
   `V9001_AQUISICAO_ALIMENTAR_DESC` (grupo/subgrupo de aquisição alimentar) —
   cobrem a mesma faixa de código validada nas dimensões de despesa (quadros
   63 a 90), relevante sobretudo para a Caderneta Coletiva.
6. **Pós-estratos, deflatores de linha de pobreza e renda líquida**: três
   fontes de referência aplicadas em qualquer tabela com a chave
   correspondente — `COD_UPA` → `COD_UPA_POS_ESTRATO` (Pós-estratos, usado na
   calibração de pesos amostrais); `UF` → `UF_DEFLATOR_LINHA_POBREZA`,
   `UF_PPP_2011`, `UF_LINHA_POBREZA_1_90` e `UF_LINHA_POBREZA_5_50` (Deflator
   de linhas de pobreza); e `V9001` → `V9001_RENDA_LIQUIDA_DESC`
   (classificação de renda líquida/dedução, mesma faixa de código do Tradutor
   de Rendimento).

7. **Medidas caseiras do Consumo Alimentar**: única fonte com chave composta
   (`V9001` + `COD_PREPARACAO_FINAL` + `COD_UNIDADE_MEDIDA_FINAL`, em vez de
   um único código) — aplicada via `merge` (não `.map()`), adiciona
   `GRAMATURA_MEDIDA_CASEIRA_REF` e `DESC_MEDIDA_CASEIRA_REF` em Consumo
   Alimentar a partir da Tabela de Medidas Caseiras do Consumo Alimentar.

As fontes 1 e 2 são combinadas em um único `DICIONARIO_POF` antes da tradução
propriamente dita, de forma que cada tabela seja percorrida uma única vez; as
fontes 3 a 7 usam colunas de saída próprias (não `<VARIAVEL>_DESC`) e são
aplicadas em etapas subsequentes. O resultado final é `TABELAS_TRADUZIDAS`, no
mesmo formato de `TABELAS_POF`, com a coluna original de cada variável sempre
preservada ao lado das colunas traduzidas.

```python
# Etapa de tradução geral: dicionário de variáveis + tabelas de referência externas,
# aplicados de uma só vez sobre todas as tabelas consolidadas em TABELAS_POF
import re

# --- Caminhos dos arquivos de referência no Google Drive -------------------
CAMINHO_DICIONARIO_POF = f"{caminho_drive}/Dicionários de variáveis.xls"
CAMINHO_ESTRATOS_POF = f"{caminho_drive}/Estratos POF 2017-2018.xls"
CAMINHO_CADASTRO_PRODUTOS = f"{caminho_drive}/Cadastro de Produtos.xls"
CAMINHO_CADASTRO_LOCAIS_AQUISICAO = f"{caminho_drive}/Cadastro de Locais de Aquisição.xls"
CAMINHO_CNAE_DOMICILIAR = f"{caminho_drive}/Atividade CNAE Domiciliar 2.0.xlsx"
CAMINHO_PRODUTOS_CONSUMO_ALIMENTAR = f"{caminho_drive}/Cadastro de Produtos do Consumo Alimentar.xls"
CAMINHO_UNIDADES_CONSUMO_ALIMENTAR = f"{caminho_drive}/Cadastro de Unidades de Medidas do Consumo Alimentar.xls"
CAMINHO_PESOS_VOLUMES = f"{caminho_drive}/Cadastro de Pesos ou Volumes.xls"
CAMINHO_OCUPACAO_COD = f"{caminho_drive}/Ocupação COD.xlsx"
CAMINHO_GRUPOS_CONSUMO_ALIMENTAR = f"{caminho_drive}/Classificação dos grupos de Consumo Alimentar.xlsx"
CAMINHO_TRADUTOR_RENDIMENTO = f"{caminho_drive}/Tradutor Rendimento.xls"
CAMINHO_DIMENSOES_DESPESA_2A5 = f"{caminho_drive}/Tradutor Despesa Perfil dimensoes 2a5.xlsx"
CAMINHO_DIMENSOES_DESPESA_6A9 = f"{caminho_drive}/Tradutor Despesa Perfil dimensoes 6a9.xlsx"
CAMINHO_TRADUTOR_ALIMENTACAO = f"{caminho_drive}/Tradutor Alimentação.xls"
CAMINHO_TRADUTOR_AQUISICAO_ALIMENTAR = f"{caminho_drive}/Tradutor Aquisição Alimentar.xls"
CAMINHO_POS_ESTRATOS_TOTAIS = f"{caminho_drive}/Pos_estratos_totais.xlsx"
CAMINHO_RENDA_LIQUIDA_DEDUCAO = f"{caminho_drive}/Renda_liquida_deducao_micro_v1.xlsx"
CAMINHO_DEFLATOR_LINHAS_POBREZA = f"{caminho_drive}/Deflator_linhas_pobreza_POF_micro_v1.xlsx"
CAMINHO_UNIDADES_MEDIDA = f"{caminho_drive}/Cadastro de Unidades de Medida.xls"
CAMINHO_MEDIDAS_CASEIRAS_CONSUMO_ALIMENTAR = f"{caminho_drive}/Tabela de Medidas Caseiras do Consumo Alimentar.xls"

# Mapeamento: nome da tabela (usado no código) -> nome da aba no dicionário de variáveis
MAPA_ABA_DICIONARIO_POF = {
    "DOMICILIO": "Domicílio",
    "MORADOR": "Morador",
    "MORADOR_QUALI_VIDA": "Morador - Qualidade de Vida",
    "ALUGUEL_ESTIMADO": "Aluguel Estimado",
    "DESPESA_COLETIVA": "Despesa Coletiva",
    "SERVICO_NAO_MONETARIO_POF2": "Serviços Não Monetários - POF 2",
    "INVENTARIO": "Inventário",
    "CADERNETA_COLETIVA": "Caderneta Coletiva",
    "DESPESA_INDIVIDUAL": "Despesa Individual",
    "SERVICO_NAO_MONETARIO_POF4": "Serviços Não Monetários - POF 4",
    "RESTRICAO_PRODUTOS_SERVICOS_SAUDE": "Restrição - Saúde",
    "RENDIMENTO_TRABALHO": "Rendimento do Trabalho",
    "OUTROS_RENDIMENTOS": "Outros Rendimentos",
    "CONDICOES_VIDA": "Condições de Vida",
    "CARACTERISTICAS_DIETA": "Características da Dieta",
    "CONSUMO_ALIMENTAR": "Consumo Alimentar",
}

# Mapeamento: nome da tabela -> {variável: nome da fonte externa (ou lista de fontes, quando
# mais de uma cobre faixas de código complementares e sem sobreposição para a mesma variável)}.
# Reflete exatamente as referências "Ver arquivo ..." descritas no dicionário de variáveis;
# V9001 em ALUGUEL_ESTIMADO fica de fora de propósito, pois o dicionário não traz nenhuma
# referência de categorias para essa variável nessa tabela.
TABELAS_FONTE_EXTERNA = {
    "DOMICILIO": {"ESTRATO_POF": "ESTRATOS"},
    "MORADOR": {"ESTRATO_POF": "ESTRATOS"},
    "MORADOR_QUALI_VIDA": {"ESTRATO_POF": "ESTRATOS"},
    "ALUGUEL_ESTIMADO": {"ESTRATO_POF": "ESTRATOS"},
    "DESPESA_COLETIVA": {"ESTRATO_POF": "ESTRATOS", "V9001": "PRODUTOS", "V9004": "LOCAL"},
    "SERVICO_NAO_MONETARIO_POF2": {"ESTRATO_POF": "ESTRATOS", "V9001": "PRODUTOS", "V9004": "LOCAL"},
    "INVENTARIO": {"ESTRATO_POF": "ESTRATOS", "V9001": "PRODUTOS"},
    "CADERNETA_COLETIVA": {
        "ESTRATO_POF": "ESTRATOS", "V9001": "PRODUTOS", "V9004": "LOCAL", "V9009": "PESOS_VOLUMES",
        "V9007": "UNIDADES_MEDIDA",
    },
    "DESPESA_INDIVIDUAL": {"ESTRATO_POF": "ESTRATOS", "V9001": "PRODUTOS", "V9004": "LOCAL"},
    "SERVICO_NAO_MONETARIO_POF4": {"ESTRATO_POF": "ESTRATOS", "V9001": "PRODUTOS", "V9004": "LOCAL"},
    "RESTRICAO_PRODUTOS_SERVICOS_SAUDE": {"ESTRATO_POF": "ESTRATOS", "V9001": "PRODUTOS"},
    "RENDIMENTO_TRABALHO": {
        "ESTRATO_POF": "ESTRATOS", "V9001": ["PRODUTOS", "RENDIMENTO"],
        "V53011": "OCUPACAO", "V53061": "CNAE",
    },
    "OUTROS_RENDIMENTOS": {"ESTRATO_POF": "ESTRATOS", "V9001": ["PRODUTOS", "RENDIMENTO"]},
    "CONDICOES_VIDA": {"ESTRATO_POF": "ESTRATOS"},
    "CARACTERISTICAS_DIETA": {"ESTRATO_POF": "ESTRATOS"},
    "CONSUMO_ALIMENTAR": {
        "ESTRATO_POF": "ESTRATOS",
        "V9001": "PRODUTOS_CONSUMO_ALIMENTAR",
        "V9007": "UNIDADES_CONSUMO_ALIMENTAR",
        "COD_UNIDADE_MEDIDA_FINAL": "UNIDADES_CONSUMO_ALIMENTAR",
    },
}

# Regex que reconhece entradas do tipo "1 – Rondônia" ou "Branco – Não Aplicável"
# na coluna "Categorias" do dicionário (exige espaço dos dois lados do traço, para
# não confundir com hifens dentro do próprio rótulo, ex.: "Segunda-Feira").
_RE_CATEGORIA = re.compile(r"^\s*(?P<codigo>\S+)\s+[-–]\s+(?P<rotulo>.+?)\s*$")

# Áreas da pesquisa, na ordem em que aparecem nas colunas da planilha de estratos
_AREAS_ESTRATOS_POF = ["Município da Capital", "Resto da Região Metropolitana", "Resto da UF", "Rural"]


def _ler_aba_dicionario(xls, nome_aba):
    """Lê uma aba do dicionário de variáveis e retorna
    {codigo_variavel: {"descricao": str, "categorias": {codigo_bruto: rotulo}}}."""
    bruto = pd.read_excel(xls, sheet_name=nome_aba, header=None)

    # O cabeçalho ("Posição Inicial", ..., "Categorias") não fica sempre na mesma linha
    linha_cabecalho = None
    for i in range(min(10, len(bruto))):
        if "Código da variável" in list(bruto.iloc[i].astype(str)):
            linha_cabecalho = i
            break
    if linha_cabecalho is None:
        print(f"Aviso: cabeçalho não encontrado na aba '{nome_aba}'. Pulando.")
        return {}

    corpo = bruto.iloc[linha_cabecalho + 1:].reset_index(drop=True)

    variaveis = {}
    variavel_atual = None
    for _, linha in corpo.iterrows():
        codigo_var, descricao, categoria = linha[3], linha[4], linha[5]

        # Cada bloco de linhas pertence à variável cujo código aparece na primeira linha do bloco
        if pd.notna(codigo_var):
            variavel_atual = str(codigo_var).strip()
            variaveis[variavel_atual] = {
                "descricao": str(descricao).strip() if pd.notna(descricao) else "",
                "categorias": {},
            }

        if variavel_atual is not None and pd.notna(categoria):
            m = _RE_CATEGORIA.match(str(categoria))
            if m:
                cod_bruto = m.group("codigo").strip()
                rotulo = m.group("rotulo").strip()
                variaveis[variavel_atual]["categorias"][cod_bruto] = rotulo

    return variaveis


def carregar_dicionario_pof(caminho_xls=CAMINHO_DICIONARIO_POF, mapa_abas=MAPA_ABA_DICIONARIO_POF):
    """Carrega o dicionário de variáveis completo: {nome_tabela: {codigo_variavel: {...}}}."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo de dicionário {caminho_completo} não encontrado. Tradução não será aplicada.")
        return {}

    xls = pd.ExcelFile(caminho_completo)
    dicionario = {}
    for nome_tabela, nome_aba in mapa_abas.items():
        if nome_aba not in xls.sheet_names:
            print(f"Aviso: aba '{nome_aba}' não encontrada para a tabela {nome_tabela}.")
            continue
        dicionario[nome_tabela] = _ler_aba_dicionario(xls, nome_aba)
    return dicionario


def _expandir_codigos_estrato(valor):
    """Converte uma célula da planilha de estratos (ex.: "1101 a 1107", "1201 e 1202",
    "1201" ou "-") na lista de códigos de ESTRATO_POF que ela representa."""
    if pd.isna(valor):
        return []
    if isinstance(valor, (int, float)):
        return [int(valor)]

    texto = str(valor).strip()
    if texto in ("-", "", "–"):
        return []

    m = re.match(r"^(\d+)\s+a\s+(\d+)$", texto)
    if m:
        inicio, fim = int(m.group(1)), int(m.group(2))
        return list(range(inicio, fim + 1))

    m = re.match(r"^(\d+)\s+e\s+(\d+)$", texto)
    if m:
        return [int(m.group(1)), int(m.group(2))]

    m = re.match(r"^(\d+)$", texto)
    if m:
        return [int(m.group(1))]

    print(f"Aviso: formato de faixa de estrato não reconhecido: {texto!r}")
    return []


def carregar_estratos_pof(caminho_xls=CAMINHO_ESTRATOS_POF, aba="estratos _Rural Total"):
    """Lê a planilha de Estratos POF (matriz UF x área) e retorna
    {codigo_estrato: "UF - Área"}."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo {caminho_completo} não encontrado. ESTRATO_POF não será traduzido.")
        return {}

    bruto = pd.read_excel(caminho_completo, sheet_name=aba, header=None)

    mapa_estratos = {}
    dentro_dos_dados = False
    for _, linha in bruto.iterrows():
        rotulo_linha = linha[0]
        resto = linha[1:5]

        if pd.isna(rotulo_linha):
            continue

        rotulo_linha = str(rotulo_linha).strip()
        if rotulo_linha.startswith("Região"):
            dentro_dos_dados = True
            continue

        if not dentro_dos_dados or resto.isna().all():
            continue

        uf_atual = rotulo_linha
        for area, valor in zip(_AREAS_ESTRATOS_POF, resto):
            for codigo in _expandir_codigos_estrato(valor):
                mapa_estratos[codigo] = f"{uf_atual} - {area}"

    return mapa_estratos


def _detectar_linha_cabecalho(bruto, marcador, n_linhas=10):
    """Procura, nas primeiras `n_linhas`, a linha em que alguma célula começa com `marcador`
    (ex.: "CÓDIGO"). Os cadastros de referência da POF nem sempre têm o cabeçalho na primeira
    linha da planilha."""
    marcador = marcador.upper()
    for i in range(min(n_linhas, len(bruto))):
        for valor in bruto.iloc[i]:
            if isinstance(valor, str) and valor.strip().upper().startswith(marcador):
                return i
    return None


def carregar_cadastro_simples(caminho_xls, aba, marcador_codigo="CÓDIGO"):
    """Lê um cadastro simples de duas colunas (código, descrição), detectando
    automaticamente a linha de cabeçalho, e retorna {codigo: descricao}."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo {caminho_completo} não encontrado.")
        return {}

    bruto = pd.read_excel(caminho_completo, sheet_name=aba, header=None)
    linha_cabecalho = _detectar_linha_cabecalho(bruto, marcador_codigo)
    if linha_cabecalho is None:
        print(f"Aviso: cabeçalho não encontrado em {caminho_completo} ({aba}).")
        return {}

    df = pd.read_excel(caminho_completo, sheet_name=aba, header=linha_cabecalho)
    col_codigo, col_descricao = df.columns[0], df.columns[1]
    codigos = pd.to_numeric(df[col_codigo], errors="coerce")
    rotulos = df[col_descricao].astype(str).str.strip()
    return {int(cod): rot for cod, rot in zip(codigos, rotulos) if pd.notna(cod)}


def carregar_cnae_domiciliar(caminho_xls=CAMINHO_CNAE_DOMICILIAR, aba="Estrutura CNAE Domiciliar 2.0"):
    """Lê a estrutura da CNAE Domiciliar 2.0 e retorna {codigo_classe: denominacao},
    usando a coluna "Classe" (5 dígitos) — a mesma granularidade de V53061."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo {caminho_completo} não encontrado. V53061 não será traduzido.")
        return {}

    bruto = pd.read_excel(caminho_completo, sheet_name=aba, header=None)
    linha_cabecalho = None
    for i in range(min(10, len(bruto))):
        if "Classe" in list(bruto.iloc[i].astype(str)):
            linha_cabecalho = i
            break
    if linha_cabecalho is None:
        print(f"Aviso: cabeçalho da CNAE Domiciliar não encontrado em {caminho_completo}.")
        return {}

    df = pd.read_excel(caminho_completo, sheet_name=aba, header=linha_cabecalho)
    classes = pd.to_numeric(df["Classe"], errors="coerce")
    rotulos = df["Denominação"].astype(str).str.strip()
    return {int(cod): rot for cod, rot in zip(classes, rotulos) if pd.notna(cod)}


def carregar_ocupacao_cod(caminho_xls=CAMINHO_OCUPACAO_COD, aba="Estrutura COD"):
    """Lê a Classificação de Ocupações para Pesquisas Domiciliares (COD) e retorna
    {codigo_grupo_base: denominacao}, usando a coluna "Grupo de base" (4 dígitos) —
    a mesma granularidade de V53011."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo {caminho_completo} não encontrado. V53011 não será traduzido.")
        return {}

    bruto = pd.read_excel(caminho_completo, sheet_name=aba, header=None)
    linha_cabecalho = None
    for i in range(min(10, len(bruto))):
        if "Grupo de base" in list(bruto.iloc[i].astype(str)):
            linha_cabecalho = i
            break
    if linha_cabecalho is None:
        print(f"Aviso: cabeçalho da Ocupação COD não encontrado em {caminho_completo}.")
        return {}

    df = pd.read_excel(caminho_completo, sheet_name=aba, header=linha_cabecalho)
    codigos = pd.to_numeric(df["Grupo de base"], errors="coerce")
    rotulos = df["Denominação"].astype(str).str.strip()
    return {int(cod): rot for cod, rot in zip(codigos, rotulos) if pd.notna(cod)}


def carregar_grupos_consumo_alimentar(caminho_xls=CAMINHO_GRUPOS_CONSUMO_ALIMENTAR, aba="Gruali"):
    """Lê a Classificação dos grupos de Consumo Alimentar (estrutura em cascata, com o
    grupo preenchido apenas na primeira linha de cada bloco) e retorna
    {codigo_item: grupo_de_alimentos}, cobrindo o mesmo código de V9001 em
    CONSUMO_ALIMENTAR (validado: 100% dos códigos aqui existem no Cadastro de Produtos
    do Consumo Alimentar)."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo {caminho_completo} não encontrado. Grupo de alimentos não será traduzido.")
        return {}

    bruto = pd.read_excel(caminho_completo, sheet_name=aba, header=None)
    linha_cabecalho = None
    for i in range(min(10, len(bruto))):
        if any(isinstance(v, str) and "Grupo de alimentos" in v for v in bruto.iloc[i]):
            linha_cabecalho = i
            break
    if linha_cabecalho is None:
        print(f"Aviso: cabeçalho dos grupos de consumo alimentar não encontrado em {caminho_completo}.")
        return {}

    df = pd.read_excel(caminho_completo, sheet_name=aba, header=linha_cabecalho)
    df.columns = [str(c).strip() for c in df.columns]
    df["Grupo de alimentos"] = df["Grupo de alimentos"].ffill()
    itens = df[df["Código do item"].notna()]
    codigos = itens["Código do item"].astype(int)
    grupos = itens["Grupo de alimentos"].astype(str).str.strip()
    return dict(zip(codigos, grupos))


_COLUNAS_NIVEL_RENDIMENTO = ["Descricao_0", "Descricao_1", "Descricao_2", "Descricao_3"]


def carregar_tradutor_rendimento(caminho_xls=CAMINHO_TRADUTOR_RENDIMENTO, aba="Rendimento 2017-2018 "):
    """Lê o Tradutor de Rendimento — classificação hierárquica (4 níveis) do código
    V9001 usado em RENDIMENTO_TRABALHO e OUTROS_RENDIMENTOS, cobrindo a faixa de
    códigos (53xxx-55xxx) que o Cadastro de Produtos geral não contempla. Retorna
    {codigo: "Nível 0 > Nível 1 > ... > Nível N"}."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo {caminho_completo} não encontrado. Não será usado para traduzir V9001.")
        return {}

    df = pd.read_excel(caminho_completo, sheet_name=aba, header=0)
    df.columns = [str(c).strip() for c in df.columns]
    codigos = pd.to_numeric(df["Codigo"], errors="coerce")

    def caminho_hierarquico(linha):
        niveis = [str(linha[c]).strip() for c in _COLUNAS_NIVEL_RENDIMENTO if pd.notna(linha[c])]
        return " > ".join(niveis)

    caminhos = df.apply(caminho_hierarquico, axis=1)
    return {int(cod): cam for cod, cam in zip(codigos, caminhos) if pd.notna(cod)}


_COLUNAS_NIVEL_DIMENSAO = ["TOTAL_DIMENSAO", "DESCRICAO_DIMENSAO", "DESCRICAO_CATEGORIA"]


def _padronizar_dimensoes(df):
    """As duas planilhas de dimensões de despesa nomeiam a coluna de código de forma
    diferente ("V9001_CODIGO_PRODUTO " com espaço, ou "CODIGO") — padroniza para
    "CODIGO" antes de concatenar."""
    df = df.copy()
    df.columns = [str(c).strip() for c in df.columns]
    col_codigo = "V9001_CODIGO_PRODUTO" if "V9001_CODIGO_PRODUTO" in df.columns else "CODIGO"
    return df.rename(columns={col_codigo: "CODIGO"})


def carregar_dimensoes_despesa(caminho_2a5=CAMINHO_DIMENSOES_DESPESA_2A5, caminho_6a9=CAMINHO_DIMENSOES_DESPESA_6A9):
    """Lê e combina as duas planilhas de dimensões de despesa (faixas de quadro 2-5 e
    6-9, sem sobreposição de código) — classificação hierárquica (3 níveis) do código
    V9001, variando conforme a variável de valor deflacionado (`VARIAVEL`, ex.:
    V8000_DEFLA, V8501_DEFLA) à qual o código está associado. Retorna
    {variavel_defla: {codigo_v9001: "Total > Dimensão > Categoria"}}."""
    caminhos = [Path(caminho_2a5), Path(caminho_6a9)]
    if not all(c.exists() for c in caminhos):
        faltantes = [str(c) for c in caminhos if not c.exists()]
        print(f"Arquivo(s) não encontrado(s): {faltantes}. Dimensões de despesa não serão traduzidas.")
        return {}

    df1 = _padronizar_dimensoes(pd.read_excel(caminho_2a5, sheet_name="DIMENSOES_DESPESA_DOIS_A_CINCO", header=0))
    df2 = _padronizar_dimensoes(pd.read_excel(caminho_6a9, sheet_name="DIMENSOES_DESPESA_SEIS_A_NOVE", header=0))
    df = pd.concat([df1, df2], ignore_index=True)
    df["CODIGO"] = pd.to_numeric(df["CODIGO"], errors="coerce")

    def caminho_hierarquico(linha):
        niveis = [str(linha[c]).strip() for c in _COLUNAS_NIVEL_DIMENSAO if pd.notna(linha[c])]
        return " > ".join(niveis)

    df["_CAMINHO"] = df.apply(caminho_hierarquico, axis=1)

    mapa_por_variavel = {}
    for variavel, grupo in df.dropna(subset=["CODIGO"]).groupby("VARIAVEL"):
        mapa_por_variavel[variavel] = dict(zip(grupo["CODIGO"].astype(int), grupo["_CAMINHO"]))
    return mapa_por_variavel


def aplicar_fonte_extra(tabelas, tabela_alvo, coluna_origem, mapa, sufixo):
    """Adiciona '<coluna_origem><sufixo>' em `tabela_alvo` a partir de `mapa`
    (código -> rótulo), sem sobrescrever a tradução '<coluna_origem>_DESC' já
    existente — útil quando uma mesma coluna de código precisa de mais de uma
    classificação simultânea (ex.: V9001 em CONSUMO_ALIMENTAR: produto x grupo
    alimentar)."""
    if tabela_alvo not in tabelas or coluna_origem not in tabelas[tabela_alvo].columns or not mapa:
        return tabelas
    resultado = dict(tabelas)
    df_novo = resultado[tabela_alvo].copy()
    df_novo[f"{coluna_origem}{sufixo}"] = df_novo[coluna_origem].map(mapa)
    resultado[tabela_alvo] = df_novo
    return resultado


def aplicar_dimensoes_despesa(tabelas, mapa_por_variavel, coluna_codigo="V9001"):
    """Para cada tabela que tenha a coluna `coluna_codigo` (V9001) e alguma das
    variáveis _DEFLA presentes em `mapa_por_variavel`, adiciona uma coluna
    '<coluna_codigo>_DIMENSAO_<VARIAVEL>' com a classificação hierárquica
    correspondente — detectado dinamicamente a partir das colunas de cada tabela,
    sem depender de uma lista fixa."""
    resultado = {}
    for nome_tabela, df in tabelas.items():
        df_novo = df.copy()
        if coluna_codigo in df_novo.columns:
            for variavel, mapa_codigos in mapa_por_variavel.items():
                if variavel in df_novo.columns:
                    df_novo[f"{coluna_codigo}_DIMENSAO_{variavel}"] = df_novo[coluna_codigo].map(mapa_codigos)
        resultado[nome_tabela] = df_novo
    return resultado


_NIVEIS_TRADUTOR_ALIMENTACAO = ["Descricao_0", "Descricao_1", "Descricao_2", "Descricao_3"]
_NIVEIS_TRADUTOR_AQUISICAO_ALIMENTAR = ["Descricao_1", "Descricao_2", "Descricao_3"]


def _carregar_mapa_hierarquico(caminho_xls, aba, colunas_nivel, col_codigo="Codigo"):
    """Lê uma planilha no formato Codigo + Nivel_N/Descricao_N e retorna
    {codigo: "Nível 0 > Nível 1 > ... > Nível N"}."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo {caminho_completo} não encontrado.")
        return {}

    df = pd.read_excel(caminho_completo, sheet_name=aba, header=0)
    df.columns = [str(c).strip() for c in df.columns]
    codigos = pd.to_numeric(df[col_codigo], errors="coerce")

    def caminho_hierarquico(linha):
        niveis = [str(linha[c]).strip() for c in colunas_nivel if c in linha and pd.notna(linha[c])]
        return " > ".join(niveis)

    caminhos = df.apply(caminho_hierarquico, axis=1)
    return {int(cod): cam for cod, cam in zip(codigos, caminhos) if pd.notna(cod)}


def carregar_tradutor_alimentacao(caminho_xls=CAMINHO_TRADUTOR_ALIMENTACAO, aba="Planilha1"):
    """Classificação de V9001 em 'Alimentação no domicílio' x 'Alimentação fora do
    domicílio', com subgrupos de alimentos (ex.: Cereais, leguminosas e oleaginosas
    > Arroz). Mesma faixa de código validada nas dimensões de despesa (quadros 63-90)."""
    return _carregar_mapa_hierarquico(caminho_xls, aba, _NIVEIS_TRADUTOR_ALIMENTACAO)


def carregar_tradutor_aquisicao_alimentar(caminho_xls=CAMINHO_TRADUTOR_AQUISICAO_ALIMENTAR, aba="Planilha1"):
    """Classificação de V9001 por grupo/subgrupo de aquisição alimentar (ex.: Cereais
    e leguminosas > Cereais > Arroz polido). Mesma faixa de código da Tradutor de
    Alimentação, com uma quebra de subgrupos diferente (mais granular no item final)."""
    return _carregar_mapa_hierarquico(caminho_xls, aba, _NIVEIS_TRADUTOR_AQUISICAO_ALIMENTAR)


def aplicar_mapa_em_todas_tabelas(tabelas, coluna_origem, mapa, sufixo):
    """Adiciona '<coluna_origem><sufixo>' em toda tabela que contenha `coluna_origem`,
    a partir de `mapa` (código -> rótulo) — para uma classificação extra que não
    depende de uma tabela específica nem de uma variável de valor deflacionado."""
    resultado = {}
    for nome_tabela, df in tabelas.items():
        df_novo = df.copy()
        if coluna_origem in df_novo.columns and mapa:
            df_novo[f"{coluna_origem}{sufixo}"] = df_novo[coluna_origem].map(mapa)
        resultado[nome_tabela] = df_novo
    return resultado


def aplicar_mapas_em_todas_tabelas(tabelas, coluna_origem, mapas_por_sufixo):
    """Aplica `aplicar_mapa_em_todas_tabelas` uma vez por (sufixo, mapa) em
    `mapas_por_sufixo` — útil para juntar várias colunas de referência de uma vez a
    partir da mesma chave (ex.: UF -> deflator, PPP, linhas de pobreza; ou
    COD_UPA -> pós-estrato, população de referência)."""
    for sufixo, mapa in mapas_por_sufixo.items():
        tabelas = aplicar_mapa_em_todas_tabelas(tabelas, coluna_origem, mapa, sufixo)
    return tabelas


def carregar_pos_estratos_totais(caminho_xls=CAMINHO_POS_ESTRATOS_TOTAIS, aba="Planilha1"):
    """Lê a tabela de pós-estratos (usada para calibração de pesos amostrais), chave
    única por COD_UPA. Retorna {"_POS_ESTRATO": {codigo_upa: pos_estrato}}."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo {caminho_completo} não encontrado. Pós-estrato não será adicionado.")
        return {}

    bruto = pd.read_excel(caminho_completo, sheet_name=aba, header=None)
    linha_cabecalho = _detectar_linha_cabecalho(bruto, "COD_UPA")
    if linha_cabecalho is None:
        print(f"Aviso: cabeçalho não encontrado em {caminho_completo}.")
        return {}

    df = pd.read_excel(caminho_completo, sheet_name=aba, header=linha_cabecalho)
    df.columns = [str(c).strip() for c in df.columns]
    col_upa = next((c for c in df.columns if c.upper().startswith("COD_UPA")), None)
    if col_upa is None:
        print(f"Aviso: coluna COD_UPA não encontrada em {caminho_completo}.")
        return {}
    codigos = pd.to_numeric(df[col_upa], errors="coerce")

    mapas = {}
    for sufixo, coluna in {"_POS_ESTRATO": "pos_estrato"}.items():
        valores = pd.to_numeric(df[coluna], errors="coerce")
        mapas[sufixo] = {int(cod): val for cod, val in zip(codigos, valores) if pd.notna(cod) and pd.notna(val)}
    return mapas


def carregar_renda_liquida_deducao(caminho_xls=CAMINHO_RENDA_LIQUIDA_DEDUCAO, aba="Renda_líquida"):
    """Lê a classificação de renda líquida/dedução de V9001 (Trabalho, Outras
    transferências, Pensão e Aposentadorias, Bolsa Família, BPC, Outras rendas) e
    retorna {codigo: "classificação > descrição"}. Mesma faixa de código validada no
    Tradutor de Rendimento (93/97 códigos em comum)."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo {caminho_completo} não encontrado. Classificação de renda líquida não será adicionada.")
        return {}

    bruto = pd.read_excel(caminho_completo, sheet_name=aba, header=None)
    linha_cabecalho = _detectar_linha_cabecalho(bruto, "Códigos")
    if linha_cabecalho is None:
        print(f"Aviso: cabeçalho não encontrado em {caminho_completo}.")
        return {}

    df = pd.read_excel(caminho_completo, sheet_name=aba, header=linha_cabecalho)
    df.columns = [str(c).strip() for c in df.columns]
    codigos = pd.to_numeric(df["Códigos"], errors="coerce")
    rotulos = df["classificação"].astype(str).str.strip() + " > " + df["Descrição"].astype(str).str.strip()
    return {int(cod): rot for cod, rot in zip(codigos, rotulos) if pd.notna(cod)}


def carregar_deflator_linhas_pobreza(caminho_xls=CAMINHO_DEFLATOR_LINHAS_POBREZA, aba="linhas e deflatores"):
    """Lê os deflatores e linhas de pobreza por UF (paridade de poder de compra e
    linhas de pobreza internacionais em reais por mês) e retorna, por sufixo de
    coluna, {codigo_uf: valor}."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo {caminho_completo} não encontrado. Deflatores de linha de pobreza não serão adicionados.")
        return {}

    df = pd.read_excel(caminho_completo, sheet_name=aba, header=0)
    df.columns = [str(c).strip() for c in df.columns]
    codigos = pd.to_numeric(df["COD_UF"], errors="coerce")

    mapas = {}
    for sufixo, coluna in {
        "_DEFLATOR_LINHA_POBREZA": "deflator_linha",
        "_PPP_2011": "PPP_2011_revisada",
        "_LINHA_POBREZA_1_90": "linha_S_1_90",
        "_LINHA_POBREZA_5_50": "linha_S_5_50",
    }.items():
        valores = pd.to_numeric(df[coluna], errors="coerce")
        mapas[sufixo] = {int(cod): val for cod, val in zip(codigos, valores) if pd.notna(cod) and pd.notna(val)}
    return mapas


def carregar_medidas_caseiras_consumo_alimentar(
    caminho_xls=CAMINHO_MEDIDAS_CASEIRAS_CONSUMO_ALIMENTAR, aba="med_cas"
):
    """Lê a Tabela de Medidas Caseiras do Consumo Alimentar (conversão de medida
    caseira para gramas) e retorna um DataFrame de referência já com as colunas
    renomeadas para bater com CONSUMO_ALIMENTAR (V9001, COD_PREPARACAO_FINAL,
    COD_UNIDADE_MEDIDA_FINAL), pronto para merge — ao contrário dos demais
    tradutores, a chave aqui é composta (item + preparação + unidade), não um único
    código, por isso a aplicação é feita via merge (`aplicar_medidas_caseiras`) em
    vez de `.map()`."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo {caminho_completo} não encontrado. Medidas caseiras não serão adicionadas.")
        return None

    df = pd.read_excel(caminho_completo, sheet_name=aba, header=0)
    df = df.rename(columns={
        "CODG_ITEM": "V9001",
        "codigo": "COD_PREPARACAO_FINAL",
        "COD_UNIDADE": "COD_UNIDADE_MEDIDA_FINAL",
        "GRAMATURA": "GRAMATURA_MEDIDA_CASEIRA_REF",
        "DESC_REF": "DESC_MEDIDA_CASEIRA_REF",
    })
    chaves = ["V9001", "COD_PREPARACAO_FINAL", "COD_UNIDADE_MEDIDA_FINAL"]
    colunas = chaves + ["GRAMATURA_MEDIDA_CASEIRA_REF", "DESC_MEDIDA_CASEIRA_REF"]
    if not all(c in df.columns for c in colunas):
        print(f"Aviso: colunas esperadas não encontradas em {caminho_completo}.")
        return None
    return df[colunas].drop_duplicates(subset=chaves)


def aplicar_medidas_caseiras(tabelas, df_referencia, tabela_alvo="CONSUMO_ALIMENTAR",
                              chaves=("V9001", "COD_PREPARACAO_FINAL", "COD_UNIDADE_MEDIDA_FINAL")):
    """Faz um left join de `df_referencia` (chave composta) em `tabela_alvo`,
    adicionando GRAMATURA_MEDIDA_CASEIRA_REF e DESC_MEDIDA_CASEIRA_REF."""
    if tabela_alvo not in tabelas or df_referencia is None or df_referencia.empty:
        return tabelas
    df = tabelas[tabela_alvo]
    if not all(c in df.columns for c in chaves):
        return tabelas
    resultado = dict(tabelas)
    resultado[tabela_alvo] = df.merge(df_referencia, on=list(chaves), how="left")
    return resultado


def _mapa_traducao(categorias):
    """Converte {codigo_bruto: rotulo} em um dict de mapeamento pronto para .map(),
    cobrindo tanto o código como string quanto como inteiro (ex.: "01" e 1),
    além de isolar o rótulo especial "Branco" (campo em branco / não aplicável)."""
    mapa = {}
    valor_branco = None
    for cod_bruto, rotulo in categorias.items():
        if cod_bruto.lower() == "branco":
            valor_branco = rotulo
            continue
        mapa[cod_bruto] = rotulo
        try:
            mapa[int(cod_bruto)] = rotulo
        except ValueError:
            pass
    return mapa, valor_branco


def traduzir_tabela(df, nome_tabela, dicionario, sufixo="_DESC"):
    """Retorna uma cópia de df com uma coluna adicional '<VARIAVEL>_DESC' para
    cada coluna categórica reconhecida no dicionário da POF."""
    if nome_tabela not in dicionario:
        print(f"Tabela {nome_tabela} não encontrada no dicionário. Pulando tradução.")
        return df

    df_traduzido = df.copy()
    variaveis = dicionario[nome_tabela]

    for coluna in df.columns:
        if coluna not in variaveis:
            continue
        categorias = variaveis[coluna]["categorias"]
        if not categorias:
            continue

        mapa, valor_branco = _mapa_traducao(categorias)
        if not mapa and valor_branco is None:
            continue

        traduzida = df[coluna].map(mapa)
        if valor_branco is not None:
            traduzida = traduzida.where(df[coluna].notna(), valor_branco)

        df_traduzido[f"{coluna}{sufixo}"] = traduzida

    return df_traduzido


# --- 1. Carrega o dicionário de variáveis (categorias inline) --------------
DICIONARIO_POF = carregar_dicionario_pof()

# --- 2. Carrega as tabelas de referência externas ---------------------------
FONTES_EXTERNAS_POF = {
    "ESTRATOS": carregar_estratos_pof(),
    "PRODUTOS": carregar_cadastro_simples(CAMINHO_CADASTRO_PRODUTOS, "T_SAS_PRODUTOS_FINAL"),
    "LOCAL": carregar_cadastro_simples(CAMINHO_CADASTRO_LOCAIS_AQUISICAO, "LOCAL"),
    "CNAE": carregar_cnae_domiciliar(),
    "PRODUTOS_CONSUMO_ALIMENTAR": carregar_cadastro_simples(CAMINHO_PRODUTOS_CONSUMO_ALIMENTAR, "X"),
    "UNIDADES_CONSUMO_ALIMENTAR": carregar_cadastro_simples(CAMINHO_UNIDADES_CONSUMO_ALIMENTAR, "POF7"),
    "PESOS_VOLUMES": carregar_cadastro_simples(CAMINHO_PESOS_VOLUMES, "POF7"),
    "OCUPACAO": carregar_ocupacao_cod(),
    "RENDIMENTO": carregar_tradutor_rendimento(),
    "UNIDADES_MEDIDA": carregar_cadastro_simples(CAMINHO_UNIDADES_MEDIDA, "POF7"),
}
for nome_fonte, mapa in FONTES_EXTERNAS_POF.items():
    print(f"Fonte externa {nome_fonte}: {len(mapa)} códigos carregados.")

# --- 3. Combina as fontes externas simples num único DICIONARIO_POF ---------
# (uma variável pode ter mais de uma fonte quando elas cobrem faixas de código
# complementares e sem sobreposição, ex.: V9001 em RENDIMENTO_TRABALHO)
for nome_tabela, variaveis_fonte in TABELAS_FONTE_EXTERNA.items():
    if nome_tabela not in DICIONARIO_POF:
        continue
    for variavel, fontes in variaveis_fonte.items():
        if variavel not in DICIONARIO_POF[nome_tabela]:
            continue
        nomes_fonte = [fontes] if isinstance(fontes, str) else fontes
        for nome_fonte in nomes_fonte:
            mapa_fonte = FONTES_EXTERNAS_POF[nome_fonte]
            DICIONARIO_POF[nome_tabela][variavel]["categorias"].update(
                {str(codigo): rotulo for codigo, rotulo in mapa_fonte.items()}
            )

# --- 4. Aplica a tradução de uma só vez sobre todas as tabelas consolidadas -
TABELAS_TRADUZIDAS = {}
for nome_tabela, df_tabela in TABELAS_POF.items():
    TABELAS_TRADUZIDAS[nome_tabela] = traduzir_tabela(df_tabela, nome_tabela, DICIONARIO_POF)
    n_colunas_traduzidas = TABELAS_TRADUZIDAS[nome_tabela].shape[1] - df_tabela.shape[1]
    print(f"{nome_tabela}: {n_colunas_traduzidas} coluna(s) _DESC adicionada(s).")

# --- 5. Grupo de alimentos (classificação extra de V9001 em CONSUMO_ALIMENTAR) ---
MAPA_GRUPOS_CONSUMO_ALIMENTAR = carregar_grupos_consumo_alimentar()
if MAPA_GRUPOS_CONSUMO_ALIMENTAR:
    TABELAS_TRADUZIDAS = aplicar_fonte_extra(
        TABELAS_TRADUZIDAS, "CONSUMO_ALIMENTAR", "V9001", MAPA_GRUPOS_CONSUMO_ALIMENTAR, "_GRUPO_DESC"
    )
    print(f"CONSUMO_ALIMENTAR: coluna V9001_GRUPO_DESC adicionada ({len(MAPA_GRUPOS_CONSUMO_ALIMENTAR)} códigos).")

# --- 6. Dimensões de despesa (classificação hierárquica extra de V9001) ---------
MAPA_DIMENSOES_DESPESA = carregar_dimensoes_despesa()
if MAPA_DIMENSOES_DESPESA:
    TABELAS_TRADUZIDAS = aplicar_dimensoes_despesa(TABELAS_TRADUZIDAS, MAPA_DIMENSOES_DESPESA)
    for nome_tabela, df_tabela in TABELAS_TRADUZIDAS.items():
        n_colunas_dimensao = sum(1 for c in df_tabela.columns if c.startswith("V9001_DIMENSAO_"))
        if n_colunas_dimensao:
            print(f"{nome_tabela}: {n_colunas_dimensao} coluna(s) de dimensão de despesa adicionada(s).")

# --- 7. Alimentação e Aquisição Alimentar (classificações extras de V9001) ------
MAPA_TRADUTOR_ALIMENTACAO = carregar_tradutor_alimentacao()
MAPA_TRADUTOR_AQUISICAO_ALIMENTAR = carregar_tradutor_aquisicao_alimentar()
TABELAS_TRADUZIDAS = aplicar_mapa_em_todas_tabelas(
    TABELAS_TRADUZIDAS, "V9001", MAPA_TRADUTOR_ALIMENTACAO, "_ALIMENTACAO_DESC"
)
TABELAS_TRADUZIDAS = aplicar_mapa_em_todas_tabelas(
    TABELAS_TRADUZIDAS, "V9001", MAPA_TRADUTOR_AQUISICAO_ALIMENTAR, "_AQUISICAO_ALIMENTAR_DESC"
)
for nome_tabela, df_tabela in TABELAS_TRADUZIDAS.items():
    n_extra = sum(1 for c in ("V9001_ALIMENTACAO_DESC", "V9001_AQUISICAO_ALIMENTAR_DESC") if c in df_tabela.columns)
    if n_extra:
        print(f"{nome_tabela}: {n_extra} coluna(s) de alimentação/aquisição alimentar adicionada(s).")

# --- 8. Pós-estratos, deflatores de linha de pobreza e renda líquida ------------
MAPAS_POS_ESTRATOS = carregar_pos_estratos_totais()
TABELAS_TRADUZIDAS = aplicar_mapas_em_todas_tabelas(TABELAS_TRADUZIDAS, "COD_UPA", MAPAS_POS_ESTRATOS)

MAPAS_DEFLATOR_POBREZA = carregar_deflator_linhas_pobreza()
TABELAS_TRADUZIDAS = aplicar_mapas_em_todas_tabelas(TABELAS_TRADUZIDAS, "UF", MAPAS_DEFLATOR_POBREZA)

MAPA_RENDA_LIQUIDA = carregar_renda_liquida_deducao()
TABELAS_TRADUZIDAS = aplicar_mapa_em_todas_tabelas(
    TABELAS_TRADUZIDAS, "V9001", MAPA_RENDA_LIQUIDA, "_RENDA_LIQUIDA_DESC"
)

SUFIXOS_PASSO_8 = (
    tuple(f"COD_UPA{s}" for s in MAPAS_POS_ESTRATOS)
    + tuple(f"UF{s}" for s in MAPAS_DEFLATOR_POBREZA)
    + ("V9001_RENDA_LIQUIDA_DESC",)
)
for nome_tabela, df_tabela in TABELAS_TRADUZIDAS.items():
    n_extra = sum(1 for c in SUFIXOS_PASSO_8 if c in df_tabela.columns)
    if n_extra:
        print(f"{nome_tabela}: {n_extra} coluna(s) de pós-estrato/pobreza/renda líquida adicionada(s).")

# --- 9. Medidas caseiras do Consumo Alimentar (chave composta: item+preparação+unidade) ---
REFERENCIA_MEDIDAS_CASEIRAS = carregar_medidas_caseiras_consumo_alimentar()
if REFERENCIA_MEDIDAS_CASEIRAS is not None:
    n_antes = TABELAS_TRADUZIDAS["CONSUMO_ALIMENTAR"].shape[1] if "CONSUMO_ALIMENTAR" in TABELAS_TRADUZIDAS else None
    TABELAS_TRADUZIDAS = aplicar_medidas_caseiras(TABELAS_TRADUZIDAS, REFERENCIA_MEDIDAS_CASEIRAS)
    if n_antes is not None:
        n_depois = TABELAS_TRADUZIDAS["CONSUMO_ALIMENTAR"].shape[1]
        print(f"CONSUMO_ALIMENTAR: {n_depois - n_antes} coluna(s) de medida caseira adicionada(s).")

# Exemplo: conferindo a tradução na tabela de Despesa Coletiva
if "DESPESA_COLETIVA" in TABELAS_TRADUZIDAS:
    colunas_exemplo = [
        "ESTRATO_POF", "ESTRATO_POF_DESC", "V9001", "V9001_DESC", "V9004", "V9004_DESC",
        "V9001_DIMENSAO_V8000_DEFLA",
    ]
    colunas_exemplo = [c for c in colunas_exemplo if c in TABELAS_TRADUZIDAS["DESPESA_COLETIVA"].columns]
    display(TABELAS_TRADUZIDAS["DESPESA_COLETIVA"][colunas_exemplo].head())

# Exemplo: conferindo a tradução na tabela de Consumo Alimentar (produto e grupo alimentar)
if "CONSUMO_ALIMENTAR" in TABELAS_TRADUZIDAS:
    colunas_exemplo = ["V9001", "V9001_DESC", "V9001_GRUPO_DESC", "V9007", "V9007_DESC"]
    colunas_exemplo = [c for c in colunas_exemplo if c in TABELAS_TRADUZIDAS["CONSUMO_ALIMENTAR"].columns]
    display(TABELAS_TRADUZIDAS["CONSUMO_ALIMENTAR"][colunas_exemplo].head())

# Exemplo: conferindo a CNAE, a ocupação (COD) e o rendimento no Rendimento do Trabalho
if "RENDIMENTO_TRABALHO" in TABELAS_TRADUZIDAS:
    colunas_exemplo = ["V53061", "V53061_DESC", "V53011", "V53011_DESC", "V9001", "V9001_DESC"]
    colunas_exemplo = [c for c in colunas_exemplo if c in TABELAS_TRADUZIDAS["RENDIMENTO_TRABALHO"].columns]
    display(TABELAS_TRADUZIDAS["RENDIMENTO_TRABALHO"][colunas_exemplo].head())

# Exemplo: conferindo a classificação de alimentação e aquisição alimentar na Caderneta Coletiva
if "CADERNETA_COLETIVA" in TABELAS_TRADUZIDAS:
    colunas_exemplo = ["V9001", "V9001_DESC", "V9001_ALIMENTACAO_DESC", "V9001_AQUISICAO_ALIMENTAR_DESC"]
    colunas_exemplo = [c for c in colunas_exemplo if c in TABELAS_TRADUZIDAS["CADERNETA_COLETIVA"].columns]
    display(TABELAS_TRADUZIDAS["CADERNETA_COLETIVA"][colunas_exemplo].head())

# Exemplo: conferindo o pós-estrato/pobreza no Domicílio e a renda líquida no Rendimento do Trabalho
if "DOMICILIO" in TABELAS_TRADUZIDAS:
    colunas_exemplo = ["UF", "COD_UPA", "COD_UPA_POS_ESTRATO", "UF_DEFLATOR_LINHA_POBREZA"]
    colunas_exemplo = [c for c in colunas_exemplo if c in TABELAS_TRADUZIDAS["DOMICILIO"].columns]
    display(TABELAS_TRADUZIDAS["DOMICILIO"][colunas_exemplo].head())

if "RENDIMENTO_TRABALHO" in TABELAS_TRADUZIDAS:
    colunas_exemplo = ["V9001", "V9001_DESC", "V9001_RENDA_LIQUIDA_DESC"]
    colunas_exemplo = [c for c in colunas_exemplo if c in TABELAS_TRADUZIDAS["RENDIMENTO_TRABALHO"].columns]
    display(TABELAS_TRADUZIDAS["RENDIMENTO_TRABALHO"][colunas_exemplo].head())

# Exemplo: conferindo a unidade de medida na Caderneta Coletiva e a medida caseira no Consumo Alimentar
if "CADERNETA_COLETIVA" in TABELAS_TRADUZIDAS:
    colunas_exemplo = ["V9007", "V9007_DESC"]
    colunas_exemplo = [c for c in colunas_exemplo if c in TABELAS_TRADUZIDAS["CADERNETA_COLETIVA"].columns]
    display(TABELAS_TRADUZIDAS["CADERNETA_COLETIVA"][colunas_exemplo].head())

if "CONSUMO_ALIMENTAR" in TABELAS_TRADUZIDAS:
    colunas_exemplo = [
        "V9001", "V9001_DESC", "COD_PREPARACAO_FINAL", "COD_UNIDADE_MEDIDA_FINAL",
        "GRAMATURA_MEDIDA_CASEIRA_REF", "DESC_MEDIDA_CASEIRA_REF",
    ]
    colunas_exemplo = [c for c in colunas_exemplo if c in TABELAS_TRADUZIDAS["CONSUMO_ALIMENTAR"].columns]
    display(TABELAS_TRADUZIDAS["CONSUMO_ALIMENTAR"][colunas_exemplo].head())
```

```python
class POFProfiler:
    def __init__(self, dict_tabelas):
        self.tabelas = {k: v for k, v in dict_tabelas.items() if isinstance(v, pd.DataFrame)}
        self.perfis_por_upa = {}

    def gerar_perfis(self, especificacoes_colunas=None):
        """
        Gera perfis e EDA para todas as tabelas divididos por COD_UPA.
        """
        for nome, df in self.tabelas.items():
            print(f"\n{'='*30} ANALISANDO TABELA: {nome} {'='*30}")

            if 'COD_UPA' not in df.columns:
                print(f"Aviso: Coluna 'COD_UPA' não encontrada em {nome}. Pulando...")
                continue

            # 1. Identificação de Tipos
            specs = especificacoes_colunas.get(nome, {}) if especificacoes_colunas else {}
            cols_numericas = []
            cols_categoricas = []

            for col in df.columns:
                if col == 'COD_UPA': continue
                tipo = specs.get(col)
                if tipo == 'numerical':
                    cols_numericas.append(col)
                elif tipo == 'categorical':
                    cols_categoricas.append(col)
                else:
                    if pd.api.types.is_numeric_dtype(df[col]) and df[col].nunique() > 20:
                        cols_numericas.append(col)
                    else:
                        cols_categoricas.append(col)

            # 2. Perfilagem Numérica Segmentada por UPA
            print(f"\n[EDA] Perfil Estatístico das Variáveis Numéricas por COD_UPA:")
            if cols_numericas:
                cols_interesse = [c for c in cols_numericas if c not in ['UF', 'ESTRATO_POF', 'NUM_DOM']]
                perfil_num = df.groupby('COD_UPA')[cols_interesse].describe().head(10)
                display(perfil_num)

            # 3. Perfilagem Categórica Segmentada por UPA
            print(f"\n[EDA] Perfil de Frequência das Variáveis Categóricas por COD_UPA:")
            if cols_categoricas:
                cat_exemplo = cols_categoricas[0]
                # Correção: cat_col -> cat_exemplo
                perfil_cat = df.groupby(['COD_UPA', cat_exemplo]).size().unstack(fill_value=0).head(10)
                print(f"Exemplo: Contagem da coluna '{cat_exemplo}' por UPA:")
                display(perfil_cat)

            # 4. Visualização de Renda/Peso por UPA (Amostra)
            col_plot = 'RENDA_TOTAL' if 'RENDA_TOTAL' in df.columns else ('PESO_FINAL' if 'PESO_FINAL' in df.columns else None)
            if col_plot:
                top_upas = df['COD_UPA'].unique()[:10]
                df_plot = df[df['COD_UPA'].isin(top_upas)]
                plt.figure(figsize=(12, 5))
                sns.barplot(x='COD_UPA', y=col_plot, data=df_plot, estimator=np.mean)
                plt.title(f"Média de {col_plot} para as primeiras 10 UPAs - {nome}")
                plt.xticks(rotation=45)
                plt.show()

            self.perfis_por_upa[nome] = {
                'total_upas': df['COD_UPA'].nunique(),
                'colunas_analisadas': cols_numericas + cols_categoricas
            }

# Execução: perfilagem sobre as tabelas já consolidadas e traduzidas em TABELAS_TRADUZIDAS
config_colunas = {"MORADOR": {"ANOS_ESTUDO": "numerical"}}
profiler = POFProfiler(TABELAS_TRADUZIDAS)
profiler.gerar_perfis(especificacoes_colunas=config_colunas)
```

## Ferramenta de Construção de Perfis

`ConstrutorPerfis` cruza variáveis (tipicamente as colunas `*_DESC` geradas na
etapa de tradução) para construir perfis descritivos a partir de
`TABELAS_TRADUZIDAS`, com três modos de uso:

- **`perfil(tabela, variaveis, ...)`** — cruza uma ou mais variáveis dentro de
  uma única tabela (ex.: UF x tipo de domicílio), retornando contagem,
  peso amostral (`PESO_FINAL`), percentual e, opcionalmente, medidas
  agregadas por combinação (média ponderada, soma, mediana ou uma função
  personalizada).
- **`perfil_cruzado(join_spec, variaveis, ...)`** — une (left join) tabelas
  diferentes pelas chaves de identificação (domicílio, unidade de consumo,
  morador) e então constrói o perfil sobre o resultado, permitindo cruzar
  variáveis de tabelas de origens distintas (ex.: tipo de domicílio x
  escolaridade do morador x renda).
- **`tabela_cruzada(tabela, var_linha, var_coluna, ...)`** — atalho para um
  pivot 2D clássico entre exatamente duas variáveis.

Também inclui `plotar` (gráfico de barras com as maiores participações de um
perfil) e `exportar` (reaproveita `exportar_para_csv` para salvar o perfil no
Drive).

```python
class ConstrutorPerfis:
    """Ferramenta para construir perfis (contagens, pesos amostrais e medidas
    agregadas) a partir das tabelas da POF já traduzidas, cruzando uma ou mais
    variáveis — dentro de uma única tabela ou entre tabelas diferentes, via join
    pelas chaves de identificação (domicílio, unidade de consumo, morador...).
    """

    def __init__(self, tabelas):
        self.tabelas = {k: v for k, v in tabelas.items() if isinstance(v, pd.DataFrame)}

    # ------------------------------------------------------------------ #
    # Perfil dentro de uma única tabela
    # ------------------------------------------------------------------ #
    def perfil(self, tabela, variaveis, coluna_peso="PESO_FINAL", medidas=None,
               ordenar_por="percentual", ascendente=False, min_contagem=1):
        """Cruza `variaveis` (lista de colunas categóricas, tipicamente as *_DESC
        geradas na etapa de tradução) e retorna uma linha por combinação observada:

          - n: número de registros da combinação;
          - peso_total: soma de `coluna_peso` (se a coluna existir na tabela);
          - percentual: participação (%) de peso_total sobre o total da tabela;
          - uma coluna por item de `medidas`, no formato
            {nome_saida: (coluna, agregador)}, em que `agregador` é "media"
            (ponderada por `coluna_peso` quando disponível), "soma", "mediana",
            ou uma função personalizada `func(sub_df) -> valor`.

        `tabela` pode ser o nome de uma tabela já registrada no construtor, ou
        diretamente um DataFrame (por exemplo, o resultado de `perfil_cruzado`).
        """
        df = self.tabelas[tabela] if isinstance(tabela, str) else tabela
        variaveis = [v for v in variaveis if v in df.columns]
        if not variaveis:
            raise ValueError("Nenhuma das variáveis informadas existe na tabela.")

        tem_peso = coluna_peso in df.columns
        peso_global = df[coluna_peso].sum() if tem_peso else None

        linhas = []
        for chave, sub_df in df.groupby(variaveis, dropna=False, observed=True):
            if len(sub_df) < min_contagem:
                continue
            chave = chave if isinstance(chave, tuple) else (chave,)
            linha = dict(zip(variaveis, chave))
            linha["n"] = len(sub_df)

            if tem_peso:
                peso_total = sub_df[coluna_peso].sum()
                linha["peso_total"] = peso_total
                linha["percentual"] = (peso_total / peso_global * 100) if peso_global else np.nan

            for nome_saida, (coluna, agregador) in (medidas or {}).items():
                linha[nome_saida] = self._agregar(sub_df, coluna, agregador, coluna_peso if tem_peso else None)

            linhas.append(linha)

        perfil_df = pd.DataFrame(linhas)
        if ordenar_por in perfil_df.columns:
            perfil_df = perfil_df.sort_values(ordenar_por, ascending=ascendente)
        return perfil_df.reset_index(drop=True)

    @staticmethod
    def _agregar(sub_df, coluna, agregador, coluna_peso):
        valores = pd.to_numeric(sub_df[coluna], errors="coerce")
        if callable(agregador):
            return agregador(sub_df)
        if agregador == "soma":
            return valores.sum()
        if agregador == "mediana":
            return valores.median()
        if agregador == "media":
            if coluna_peso and coluna_peso in sub_df.columns:
                pesos = pd.to_numeric(sub_df[coluna_peso], errors="coerce")
                mascara = valores.notna() & pesos.notna()
                if mascara.sum() == 0 or pesos[mascara].sum() == 0:
                    return np.nan
                return float(np.average(valores[mascara], weights=pesos[mascara]))
            return valores.mean()
        raise ValueError(f"Agregador desconhecido: {agregador!r}")

    # ------------------------------------------------------------------ #
    # Perfil cruzando tabelas diferentes (join + perfil)
    # ------------------------------------------------------------------ #
    def perfil_cruzado(self, join_spec, variaveis, chaves_padrao=("UF", "COD_UPA", "NUM_DOM"),
                        coluna_peso="PESO_FINAL", medidas=None, **kwargs):
        """Une (left join) as tabelas indicadas em `join_spec`, na ordem dada, e então
        constrói o perfil sobre o resultado — permitindo cruzar variáveis de tabelas
        de origens diferentes (ex.: tipo de domicílio x escolaridade do morador).

        Cada item de `join_spec` é o nome de uma tabela (usa `chaves_padrao` como
        chave de junção) ou uma tupla `(nome_tabela, chaves)` para uma chave
        específica — necessário, por exemplo, ao unir tabelas em nível de morador ou
        de despesa (chaves incluindo NUM_UC e/ou COD_INFORMANTE) em vez do nível de
        domicílio. Atenção ao escolher `coluna_peso` após o join: colunas repetidas
        entre as tabelas (como PESO_FINAL) recebem o sufixo `_<nome_tabela>` a partir
        da segunda tabela unida — a coluna sem sufixo é sempre a da primeira tabela.
        """
        base = None
        for item in join_spec:
            nome_tabela, chaves = item if isinstance(item, tuple) else (item, chaves_padrao)
            df = self.tabelas[nome_tabela]
            chaves = [c for c in chaves if c in df.columns]
            if base is None:
                base = df
                continue
            chaves_comuns = [c for c in chaves if c in base.columns]
            if not chaves_comuns:
                raise ValueError(f"Nenhuma chave em comum entre a base atual e '{nome_tabela}'.")
            base = base.merge(df, on=chaves_comuns, how="left", suffixes=("", f"_{nome_tabela}"))

        return self.perfil(base, variaveis, coluna_peso=coluna_peso, medidas=medidas, **kwargs)

    # ------------------------------------------------------------------ #
    # Tabela cruzada (pivot) para exatamente duas variáveis
    # ------------------------------------------------------------------ #
    def tabela_cruzada(self, tabela, variavel_linha, variavel_coluna, coluna_peso="PESO_FINAL",
                        valores="percentual"):
        """Pivot 2D clássico: `variavel_linha` nas linhas, `variavel_coluna` nas
        colunas, com `valores` em {"n", "peso_total", "percentual"} preenchendo as
        células (0 onde não há observações)."""
        perfil_df = self.perfil(tabela, [variavel_linha, variavel_coluna], coluna_peso=coluna_peso)
        if valores not in perfil_df.columns:
            raise ValueError(f"'{valores}' não disponível (a tabela tem a coluna de peso informada?).")
        return perfil_df.pivot(index=variavel_linha, columns=variavel_coluna, values=valores).fillna(0)

    # ------------------------------------------------------------------ #
    # Visualização e exportação
    # ------------------------------------------------------------------ #
    def plotar(self, perfil_df, variaveis, valor="percentual", top_n=15, titulo=None):
        """Gráfico de barras horizontais com as `top_n` combinações de maior `valor`
        em `perfil_df` (o resultado de `perfil`/`perfil_cruzado`), rotuladas pela
        concatenação das colunas em `variaveis`."""
        dados = perfil_df.sort_values(valor, ascending=False).head(top_n).iloc[::-1]
        rotulos = dados[variaveis].astype(str).agg(" | ".join, axis=1)

        plt.figure(figsize=(10, max(3, 0.4 * len(dados))))
        plt.barh(rotulos, dados[valor])
        plt.xlabel(valor)
        plt.title(titulo or f"Perfil por {' x '.join(variaveis)}")
        plt.tight_layout()
        plt.show()

    def exportar(self, perfil_df, nome_arquivo):
        """Exporta um perfil já construído para CSV, reaproveitando `exportar_para_csv`."""
        exportar_para_csv(perfil_df, nome_arquivo)


# Exemplo de uso sobre as tabelas consolidadas e traduzidas
construtor_perfis = ConstrutorPerfis(TABELAS_TRADUZIDAS)

if "DOMICILIO" in construtor_perfis.tabelas:
    variaveis_domicilio = [c for c in ["UF_DESC", "V0201_DESC"] if c in construtor_perfis.tabelas["DOMICILIO"].columns]
    if len(variaveis_domicilio) >= 2:
        print("Perfil: UF x tipo de domicílio")
        display(construtor_perfis.perfil("DOMICILIO", variaveis_domicilio).head(10))

if "MORADOR" in construtor_perfis.tabelas:
    variaveis_morador = [
        c for c in ["V0401_DESC", "NIVEL_INSTRUCAO_DESC"] if c in construtor_perfis.tabelas["MORADOR"].columns
    ]
    if len(variaveis_morador) >= 2 and "RENDA_TOTAL" in construtor_perfis.tabelas["MORADOR"].columns:
        print("\nPerfil: sexo x nível de instrução do morador, com renda média ponderada")
        perfil_morador = construtor_perfis.perfil(
            "MORADOR", variaveis_morador, medidas={"RENDA_MEDIA": ("RENDA_TOTAL", "media")}
        )
        display(perfil_morador.head(10))

if "DOMICILIO" in construtor_perfis.tabelas and "MORADOR" in construtor_perfis.tabelas:
    variaveis_cruzadas = [
        c for c in ["V0201_DESC", "V0401_DESC"]
        if c in construtor_perfis.tabelas["DOMICILIO"].columns or c in construtor_perfis.tabelas["MORADOR"].columns
    ]
    if len(variaveis_cruzadas) == 2:
        print("\nPerfil cruzado entre tabelas: tipo de domicílio x sexo do morador")
        display(
            construtor_perfis.perfil_cruzado(["DOMICILIO", "MORADOR"], variaveis_cruzadas).head(10)
        )
```
