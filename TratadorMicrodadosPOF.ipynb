## Etapa de Tradução das Tabelas

Esta etapa traduz os códigos numéricos das variáveis categóricas de cada tabela
para seus rótulos descritivos, seguindo o "Dicionário de variáveis" oficial da POF
(planilha em anexo). Para cada variável categórica reconhecida, é criada uma nova
coluna `<VARIAVEL>_DESC` com o texto correspondente ao código, mantendo a coluna
original intacta (necessária para cálculos e junções entre tabelas).

```python
# Etapa adicional: Tradução das tabelas com base no Dicionário de variáveis da POF (.xls)
import re

# Caminho do dicionário de variáveis no Google Drive (mesma pasta dos microdados)
CAMINHO_DICIONARIO_POF = f"{caminho_drive}/Dicionários de variáveis.xls"

# Mapeamento: nome da tabela (usado no código) -> nome da aba no dicionário .xls
MAPA_ABA_DICIONARIO_POF = {
    "DOMICILIO": "Domicílio",
    "MORADOR": "Morador",
    "MORADOR_QUALI_VIDA": "Morador - Qualidade de Vida",
    "ALUGUEL_ESTIMADO": "Aluguel Estimado",
    "DESPESA_COLETIVA": "Despesa Coletiva",
    "SERVICO_NAO_MONETARIO_POF2": "Serviços Não Monetários - POF 2",
    "INVENTARIO": "Inventário",
    "CADERNETA_COLETIVA": "Caderneta Coletiva",
    "DESPESA_INDIVIDUAL": "Despesa Individual",
    "SERVICO_NAO_MONETARIO_POF4": "Serviços Não Monetários - POF 4",
    "RESTRICAO_PRODUTOS_SERVICOS_SAUDE": "Restrição - Saúde",
    "RENDIMENTO_TRABALHO": "Rendimento do Trabalho",
    "OUTROS_RENDIMENTOS": "Outros Rendimentos",
    "CONDICOES_VIDA": "Condições de Vida",
    "CARACTERISTICAS_DIETA": "Características da Dieta",
    "CONSUMO_ALIMENTAR": "Consumo Alimentar",
}

# Regex que reconhece entradas do tipo "1 – Rondônia" ou "Branco – Não Aplicável"
# na coluna "Categorias" do dicionário (exige espaço dos dois lados do traço, para
# não confundir com hifens dentro do próprio rótulo, ex.: "Segunda-Feira").
_RE_CATEGORIA = re.compile(r"^\s*(?P<codigo>\S+)\s+[-–]\s+(?P<rotulo>.+?)\s*$")


def _ler_aba_dicionario(xls, nome_aba):
    """Lê uma aba do dicionário de variáveis e retorna
    {codigo_variavel: {"descricao": str, "categorias": {codigo_bruto: rotulo}}}."""
    bruto = pd.read_excel(xls, sheet_name=nome_aba, header=None)

    # O cabeçalho ("Posição Inicial", ..., "Categorias") não fica sempre na mesma linha
    linha_cabecalho = None
    for i in range(min(10, len(bruto))):
        if "Código da variável" in list(bruto.iloc[i].astype(str)):
            linha_cabecalho = i
            break
    if linha_cabecalho is None:
        print(f"Aviso: cabeçalho não encontrado na aba '{nome_aba}'. Pulando.")
        return {}

    corpo = bruto.iloc[linha_cabecalho + 1:].reset_index(drop=True)

    variaveis = {}
    variavel_atual = None
    for _, linha in corpo.iterrows():
        codigo_var, descricao, categoria = linha[3], linha[4], linha[5]

        # Cada bloco de linhas pertence à variável cujo código aparece na primeira linha do bloco
        if pd.notna(codigo_var):
            variavel_atual = str(codigo_var).strip()
            variaveis[variavel_atual] = {
                "descricao": str(descricao).strip() if pd.notna(descricao) else "",
                "categorias": {},
            }

        if variavel_atual is not None and pd.notna(categoria):
            m = _RE_CATEGORIA.match(str(categoria))
            if m:
                cod_bruto = m.group("codigo").strip()
                rotulo = m.group("rotulo").strip()
                variaveis[variavel_atual]["categorias"][cod_bruto] = rotulo

    return variaveis


def carregar_dicionario_pof(caminho_xls=CAMINHO_DICIONARIO_POF, mapa_abas=MAPA_ABA_DICIONARIO_POF):
    """Carrega o dicionário de variáveis completo: {nome_tabela: {codigo_variavel: {...}}}."""
    caminho_completo = Path(caminho_xls)
    if not caminho_completo.exists():
        print(f"Arquivo de dicionário {caminho_completo} não encontrado. Tradução não será aplicada.")
        return {}

    xls = pd.ExcelFile(caminho_completo)
    dicionario = {}
    for nome_tabela, nome_aba in mapa_abas.items():
        if nome_aba not in xls.sheet_names:
            print(f"Aviso: aba '{nome_aba}' não encontrada para a tabela {nome_tabela}.")
            continue
        dicionario[nome_tabela] = _ler_aba_dicionario(xls, nome_aba)
    return dicionario


def _mapa_traducao(categorias):
    """Converte {codigo_bruto: rotulo} em um dict de mapeamento pronto para .map(),
    cobrindo tanto o código como string quanto como inteiro (ex.: "01" e 1),
    além de isolar o rótulo especial "Branco" (campo em branco / não aplicável)."""
    mapa = {}
    valor_branco = None
    for cod_bruto, rotulo in categorias.items():
        if cod_bruto.lower() == "branco":
            valor_branco = rotulo
            continue
        mapa[cod_bruto] = rotulo
        try:
            mapa[int(cod_bruto)] = rotulo
        except ValueError:
            pass
    return mapa, valor_branco


def traduzir_tabela(df, nome_tabela, dicionario, sufixo="_DESC"):
    """Retorna uma cópia de df com uma coluna adicional '<VARIAVEL>_DESC' para
    cada coluna categórica reconhecida no dicionário da POF."""
    if nome_tabela not in dicionario:
        print(f"Tabela {nome_tabela} não encontrada no dicionário. Pulando tradução.")
        return df

    df_traduzido = df.copy()
    variaveis = dicionario[nome_tabela]

    for coluna in df.columns:
        if coluna not in variaveis:
            continue
        categorias = variaveis[coluna]["categorias"]
        if not categorias:
            continue

        mapa, valor_branco = _mapa_traducao(categorias)
        if not mapa and valor_branco is None:
            continue

        traduzida = df[coluna].map(mapa)
        if valor_branco is not None:
            traduzida = traduzida.where(df[coluna].notna(), valor_branco)

        df_traduzido[f"{coluna}{sufixo}"] = traduzida

    return df_traduzido


# Carrega o dicionário de variáveis
DICIONARIO_POF = carregar_dicionario_pof()

# Aplica a tradução em todas as tabelas carregadas na etapa anterior
TABELAS_TRADUZIDAS = {}
for nome_tabela in MAPA_ABA_DICIONARIO_POF:
    df_tabela = globals().get(nome_tabela)
    if isinstance(df_tabela, pd.DataFrame):
        TABELAS_TRADUZIDAS[nome_tabela] = traduzir_tabela(df_tabela, nome_tabela, DICIONARIO_POF)
        n_colunas_traduzidas = TABELAS_TRADUZIDAS[nome_tabela].shape[1] - df_tabela.shape[1]
        print(f"{nome_tabela}: {n_colunas_traduzidas} coluna(s) traduzida(s) adicionada(s).")
    else:
        print(f"{nome_tabela}: tabela não carregada, tradução ignorada.")

# Exemplo: conferindo a tradução da tabela de Domicílios
if "DOMICILIO" in TABELAS_TRADUZIDAS:
    colunas_desc = [c for c in TABELAS_TRADUZIDAS["DOMICILIO"].columns if c.endswith("_DESC")]
    display(TABELAS_TRADUZIDAS["DOMICILIO"][["UF"] + colunas_desc[:3]].head())
```

```python
class POFProfiler:
    def __init__(self, dict_tabelas):
        self.tabelas = {k: v for k, v in dict_tabelas.items() if isinstance(v, pd.DataFrame)}
        self.perfis_por_upa = {}

    def gerar_perfis(self, especificacoes_colunas=None):
        """
        Gera perfis e EDA para todas as tabelas divididos por COD_UPA.
        """
        for nome, df in self.tabelas.items():
            print(f"\n{'='*30} ANALISANDO TABELA: {nome} {'='*30}")

            if 'COD_UPA' not in df.columns:
                print(f"Aviso: Coluna 'COD_UPA' não encontrada em {nome}. Pulando...")
                continue

            # 1. Identificação de Tipos
            specs = especificacoes_colunas.get(nome, {}) if especificacoes_colunas else {}
            cols_numericas = []
            cols_categoricas = []

            for col in df.columns:
                if col == 'COD_UPA': continue
                tipo = specs.get(col)
                if tipo == 'numerical':
                    cols_numericas.append(col)
                elif tipo == 'categorical':
                    cols_categoricas.append(col)
                else:
                    if pd.api.types.is_numeric_dtype(df[col]) and df[col].nunique() > 20:
                        cols_numericas.append(col)
                    else:
                        cols_categoricas.append(col)

            # 2. Perfilagem Numérica Segmentada por UPA
            print(f"\n[EDA] Perfil Estatístico das Variáveis Numéricas por COD_UPA:")
            if cols_numericas:
                cols_interesse = [c for c in cols_numericas if c not in ['UF', 'ESTRATO_POF', 'NUM_DOM']]
                perfil_num = df.groupby('COD_UPA')[cols_interesse].describe().head(10)
                display(perfil_num)

            # 3. Perfilagem Categórica Segmentada por UPA
            print(f"\n[EDA] Perfil de Frequência das Variáveis Categóricas por COD_UPA:")
            if cols_categoricas:
                cat_exemplo = cols_categoricas[0]
                # Correção: cat_col -> cat_exemplo
                perfil_cat = df.groupby(['COD_UPA', cat_exemplo]).size().unstack(fill_value=0).head(10)
                print(f"Exemplo: Contagem da coluna '{cat_exemplo}' por UPA:")
                display(perfil_cat)

            # 4. Visualização de Renda/Peso por UPA (Amostra)
            col_plot = 'RENDA_TOTAL' if 'RENDA_TOTAL' in df.columns else ('PESO_FINAL' if 'PESO_FINAL' in df.columns else None)
            if col_plot:
                top_upas = df['COD_UPA'].unique()[:10]
                df_plot = df[df['COD_UPA'].isin(top_upas)]
                plt.figure(figsize=(12, 5))
                sns.barplot(x='COD_UPA', y=col_plot, data=df_plot, estimator=np.mean)
                plt.title(f"Média de {col_plot} para as primeiras 10 UPAs - {nome}")
                plt.xticks(rotation=45)
                plt.show()

            self.perfis_por_upa[nome] = {
                'total_upas': df['COD_UPA'].nunique(),
                'colunas_analisadas': cols_numericas + cols_categoricas
            }

# Execução Atualizada
tabelas_disponiveis = {
    "MORADOR": globals().get("MORADOR"),
    "DOMICILIO": globals().get("DOMICILIO"),
    "RENDIMENTO_TRABALHO": globals().get("RENDIMENTO_TRABALHO"),
    "CONSUMO_ALIMENTAR": globals().get("CONSUMO_ALIMENTAR")
}

config_colunas = {"MORADOR": {"ANOS_ESTUDO": "numerical"}}
profiler = POFProfiler(tabelas_disponiveis)
profiler.gerar_perfis(especificacoes_colunas=config_colunas)
```
